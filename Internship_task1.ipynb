```python
import pandas as pd
import numpy as np
from google.colab import files
import os

# 1️⃣ Load Dataset
df = pd.read_csv("/content/sales_data_sample.csv", encoding='latin1')
print("✅ Data Loaded Successfully")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
print("\nInitial Preview:")
print(df.head())

# 2️⃣ Check Missing Values
print("\n🔍 Missing Values Before Cleaning:")
print(df.isnull().sum())

# Example handling missing values
df = df.fillna({
    'STATE': 'Unknown',
    'POSTALCODE': 0,
    'TERRITORY': 'Not Assigned'
})

# Drop rows if essential columns are missing
df = df.dropna(subset=['ORDERNUMBER', 'CUSTOMERNAME'])

# 3️⃣ Remove Duplicate Rows
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]
print(f"\n🧾 Duplicates Removed: {before - after}")

# 4️⃣ Standardize Text Columns
text_cols = ['STATUS', 'PRODUCTLINE', 'COUNTRY', 'DEALSIZE']
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.title()

# 5️⃣ Convert Date Columns
if 'ORDERDATE' in df.columns:
    df['ORDERDATE'] = pd.to_datetime(df['ORDERDATE'], errors='coerce')
    df['ORDERDATE'] = df['ORDERDATE'].dt.strftime('%d-%m-%Y')

# 6️⃣ Clean Column Names
df.columns = (
    df.columns.str.strip()
              .str.lower()
              .str.replace(' ', '_')
              .str.replace('[^a-zA-Z0-9_]', '', regex=True)
)

# 7️⃣ Fix Data Types
num_cols = ['quantityordered', 'priceeach', 'sales', 'postalcode']
for col in num_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Ensure postal code is integer if numeric
if 'postalcode' in df.columns:
    df['postalcode'] = df['postalcode'].astype('Int64')

# 8️⃣ Summary of Changes
summary = {
    "Missing values handled": True,
    "Duplicates removed": True,
    "Text standardized": True,
    "Dates formatted": True,
    "Column names cleaned": True,
    "Data types fixed": True
}

print("\n✅ Data Cleaning Summary:")
for k, v in summary.items():
    print(f" - {k}: {v}")

# 9️⃣ Save Cleaned Dataset
# Create the directory if it doesn't exist
os.makedirs('cleaned_data', exist_ok=True)
df.to_csv("cleaned_data/sales_data_cleaned.csv", index=False)
print("\n💾 Cleaned dataset saved as 'cleaned_data/sales_data_cleaned.csv'")
files.download("cleaned_data/sales_data_cleaned.csv")


# 🔟 Quick Validation
print("\nCleaned Data Preview:")
print(df.head())
print("\nFinal Shape:", df.shape)
```
